```python
import pandas as pd
import numpy as np

matches = pd.read_csv('data/matches.csv')
matches.head(15)
```

```python
print(matches.columns)
```

```python
df = matches[['team1', 'team2', 'venue', 'toss_winner', 'toss_decision', 'winner']]
df = df.dropna()
print(df.shape)
df.head()
```

```python
#Encoding - before data cleaning, to see what needs to be corrected

from sklearn.preprocessing import LabelEncoder

# Encode toss_decision
le = LabelEncoder()
df['toss_decision'] = le.fit_transform(df['toss_decision'])

# Encode winner (target)
df['winner_encoded'] = le.fit_transform(df['winner'])

# One Hot Encode the rest
df_encoded = pd.get_dummies(df, columns=['team1', 'team2', 'venue', 'toss_winner'])

print(df_encoded.shape)
df_encoded.head()
```

```python
#which team got assigned with which number?

le_winner = LabelEncoder()
le_winner.fit(df['winner'])
for i, team in enumerate(le_winner.classes_):
    print(i, team)
```
